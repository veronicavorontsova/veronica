# Seaborn

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset('penguins')
```

```python
df = df.dropna()
```

```python
df.shape
```

```python
# there are 333 rows and 7 columns
```

```python
df.head()
```

```python
df.tail()
```

```python
# what each column represents:
# species: the species of the penguin
# island: the island that the penguin lives on
# bill_length_mm: the length of the penguins bill in millimeters
# bill_depth_mm: the depth of the penguins bill in millimeters
# flipper_length_mm: the length of the penguins flippers in millimeters
# body_mass_g: the penguins body mass in grams 
# sex: if the penguin is male or female 
```

```python
fig, axes = plt.subplots(2, 2, figsize=(15, 15))
sns.scatterplot(data = df, x="bill_length_mm", y="bill_depth_mm", color="purple",ax=axes[0,0])
axes[0, 0].set_title("Bill Length vs. Bill Depth")
axes[0, 0].set_xlabel("Bill Length in millimeters")
axes[0, 0].set_ylabel("Bill Depth in millimeters")
axes[0,0].legend(["Penguin"])
sns.scatterplot(data = df, x="flipper_length_mm", y="body_mass_g", color="green",ax=axes[0,1])
axes[0, 1].set_title("Flipper Length vs. Body Mass")
axes[0, 1].set_xlabel("Flipper Length in millimeters")
axes[0, 1].set_ylabel("Body Mass in grams")
sns.histplot(data = df, x="flipper_length_mm", color="red", ax=axes[1, 0])
axes[1,0].set_title("Distribution of Flipper Length")
axes[1, 0].set_xlabel("Flipper Length in millimeters")
sns.boxplot(data = df, x="island", y="body_mass_g", color="coral", ax=axes[1, 1])
axes[1,1].set_title("Body Mass by Island")
axes[1,1].set_xlabel("Island")
axes[1,1].set_ylabel("Body Mass in grams")
plt.show()
```

```python
# The top left scatter plot shows the association between bill length (mm) and bill depth (mm) for penguins.
# The top right scatter plot shows the association between flipper length (mm) and body mass (g) for penguins. 
# The bottom left histogram shows the distribution of flipper Lengths (mm) of the penguins.
# The bottom right boxplots shows the distribution of body mass (g) for penguins on each island. 
# I learned that there is a very weak relationship between bill length (mm) and bill depth (mm) for penguins. 
# There is a moderately strong and positive relationship between flipper length (mm) and body mass (g) for penguins. 
# The majority of penguins have a flipper length between 185 and 200 millimeters. 
# Biscoe Island has more variability in the body mass (g) of penguins than the other islands. 
```
