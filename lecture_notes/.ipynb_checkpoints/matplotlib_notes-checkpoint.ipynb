```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# sin function

x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)
```

```python
plt.figure(figsize = (6, 4))
plt.plot(x,y)
plt.show()
```

```python
y1 = np.sin(x)
y2 = np.cos(x)
```

```python
plt.figure()
plt.scatter(x, y1, label = "sin(x)", color = "green")
plt.scatter(x, y2, label = "cos(x)", color = "pink")

plt.title("sine and cosine wave") 
plt.xlabel("x values (radians)")
plt.ylabel("f(x)")

plt.legend()
plt.show()
```

```python
# customization
plt.figure()
plt.plot(x, y1, color = "xkcd:icky green", linestyle = "-.", linewidth = 3, alpha = 0.5)

plt.show()
```

```python
plt.figure()
plt.plot(x, y1, color = "xkcd:icky green", marker = "*")

plt.show()
```
