# 3.1 1D NumPy Arrays 

```python
list = [20, 27, 9, 12, 31, 17, 4, 50, 39, 35]
```

```python
arr = np.array(list)
arr
```

```python
arr * 4
```

```python
np.mean(arr)
```

```python
np.median(arr)
```

```python
np.std(arr)
```

```python
np.argmin(arr)
```

```python
np.argmax(arr) 
```

```python
arr[6]
```

```python
arr[7]
```

```python
np.min(arr)
```

```python
np.max(arr)
```

```python
# yes they match from the previous step 
```

```python
new_array = arr[(arr > 20)]
new_array
```

# 3.2 2D NumPy Arrays 

```python
matrix = [
    [5, 12, 17, 9],
    [8, 21, 14, 3],
    [6, 13, 19, 11]
] 
```

```python
matrix_array = np.array(matrix)
matrix_array
```

```python
matrix_array.shape
```

```python
# 3 rows and 4 columns
```

```python
matrix_array[1]
```
