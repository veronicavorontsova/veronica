```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    return A * np.sin(w * x)
```

```python
x = np.linspace(0, 2*np.pi, 1000)
```

```python
amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
```

```python
fig, axes = plt.subplots(1, 3, figsize=(25, 5))
for A, w in zip(amplitude, frequency):
    axes[0].plot(x, make_sine_wave(x, A, 1), linestyle='--', label = f"A = {A}")
    axes[0].set_title("Varying the amplitude in a sine wave", fontsize = 18)
    axes[0].set_xlabel("x (radians)", fontsize = 18)
    axes[0].set_ylabel("y", fontsize = 18)
    axes[0].legend()
    axes[1].plot(x, make_sine_wave(x, 1, w), linestyle=':', label = f"ω = {w}")
    axes[1].set_title("Varying the frequency in a sine wave", fontsize = 18)
    axes[1].set_xlabel("x (radians)", fontsize = 18)
    axes[1].set_ylabel("y", fontsize = 18)
    axes[1].legend()
    axes[2].plot(x, make_sine_wave(x, A, w), linestyle='-.', label = f"A = {A}, ω = {w}")
    axes[2].set_title("Varying both the amplitude and frequency in a sine wave", fontsize = 18)
    axes[2].set_xlabel("x (radians)", fontsize = 18)
    axes[2].set_ylabel("y", fontsize = 18)
    axes[2].legend()
plt.show()
```
