```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt",
skiprows=1)
```

```python
data.shape
```

```python
# There are 2642 rows and 2 columns
```

```python
magnitude = data[:, 0]
```

```python
temperature = data[:, 1]
```

```python
print(np.min(magnitude))
```

```python
print(np.argmin(magnitude))
```

```python
# The star (index 1988) with the lowest magnitude is the brightest star
```

```python
print(np.max(magnitude))
```

```python
print(np.argmax(magnitude))
```

```python
# The star (index 264) with the highest magnitude is the dimmest star
```

```python
print(np.min(temperature))
```

```python
print(np.argmin(temperature))
```

```python
# The star (index 1923) is the highest temperature is the hottest 
```

```python
print(np.max(temperature))
```

```python
print(np.argmax(temperature))
```

```python
# The star (index 2613) with the lowest temperature is the coldest 
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
```

```python
plt.figure(figsize = (8,6))
plt.scatter(x=log_temperature, y=log_magnitude, color="pink", label="stars")
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.scatter(x=log_temperature[1988], y=log_magnitude[1988], color="black", marker="*", s=100, label="brightest star")
plt.scatter(x=log_temperature[264], y=log_magnitude[264], color="purple", marker="P", s=100, label="dimmest star")
plt.scatter(x=log_temperature[1923], y=log_magnitude[1923], color="green", marker="v", s=100, label="hottest star")
plt.scatter(x=log_temperature[2613], y=log_magnitude[2613], color="orange", marker="h", s=100, label="coldest star")

plt.legend()
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
plt.colorbar(label="Temperature (K)", orientation="horizontal").ax.invert_xaxis()
plt.show()
```
