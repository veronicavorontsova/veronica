```python
import numpy as np
```

```python
list = [20, 40, 60, 80, 100]

list_divided = []
for num in list:
    list_divided.append(num/5)

print(list_divided)
```

```python
array = np.array([20, 40, 60, 80, 100])
print(array/5)
```

```python
nested_list = [
    [20, 40],
    [60, 80],
    [100, 120]
]
print(nested_list[1][0])

ndarray = np.array([
    [20, 40],
    [60, 80],
    [100, 120]
])
print(ndarray[1, 0])
```
