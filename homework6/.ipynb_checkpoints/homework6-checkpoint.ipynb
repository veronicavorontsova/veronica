# 4 Prime Star Systems 

```python
def row_at_least_one_prime(array):
    def prime_or_not(num):
        if num < 2:
            return False
        for i in range(2, int(num**0.5) + 1):
            if num%i == 0:
                return False 
            else:
                return True 
    new_array = []
    for i in array:
        for j in i:
            if prime_or_not(j) == True: 
                new_array.append(i.tolist())
                break 
    return new_array     
```

```python
arr = np.array([
    [2, 3, 5],
    [14, 6, 8],
    [11, 13, 17],
    [7, 10, 13],
])
```

```python
row_at_least_one_prime(arr)
```

# 5 Let's Play Checkers!

```python
def eight_by_eight():    
    return np.zeros((8, 8))
```

```python
eight = eight_by_eight()
eight
```

```python
eight[::2, ::2] = 1
eight
```

```python
eight[1::2, 1::2] = 1
eight
```

```python
def checkerboard():   
    checkerboard = np.zeros((8, 8))
    checkerboard[::2, ::2] = 1
    checkerboard[1::2, 1::2] = 1
    return checkerboard
```

```python
checkerboard()
```

```python
def reverse_checkerboard():
    reversed_checkerboard = np.zeros((8, 8))
    reversed_checkerboard[1::2, ::2] = 1
    reversed_checkerboard[::2, 1::2] = 1
    return reversed_checkerboard
```

```python
reverse_checkerboard()
```

# 6 The Expanding Universe 

```python
def expansion(array, space):
    spaces = " " * space
    result = []
    for i in array: 
        expansion = spaces.join(i)
        result = np.append(result, expansion) 
    return np.array(result)
```

```python
universe = np.array(['galaxy', 'clusters'])
```

```python
expansion(universe, 1)
```

```python
expansion(universe, 2)
```

# 7 Second-Brightest Star 

```python
def secondDimmest(array):
    second_dimmest = []
    for i in array:
        row = np.array(i)
        dimmest = np.min(row)
        row_removed = row[row != dimmest]
        second_dimm = np.min(row_removed)
        second_dimmest = np.append(second_dimmest, second_dimm)
    return second_dimmest
```

```python
np.random.seed(123)
```

```python
stars = np.random.randint(500, 2000, (5, 5))
stars
```

```python
secondDimmest(stars)
```
